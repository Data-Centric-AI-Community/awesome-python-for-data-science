# Missing Data Imputation using Machine Learning Methods

##  Introduction

Missing data is a common problem in real-world datasets. Machine learning models cannot handle missing values directly, so we need to fill (impute) them.

In this tutorial, we will explore different techniques to handle missing data.

##  Why Handle Missing Data?

- Many ML algorithms cannot work with missing values
- Missing data can reduce model accuracy
- Proper imputation improves model performance

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer, KNNImputer
import matplotlib.pyplot as plt
```

```python
data = {
    "Age": [25, 30, np.nan, 35, 40, np.nan],
    "Salary": [50000, 60000, 55000, np.nan, 65000, 70000]
}

df = pd.DataFrame(data)
df
```

##  Dataset

We created a small dataset with missing values to demonstrate imputation techniques.

```python
# Check missing values
df.isnull().sum()
```

```python
df.isnull().sum().plot(kind='bar')
plt.title("Missing Values Count")
plt.show()
```

# METHODS


## 🔹 Mean Imputation

Missing values are replaced with the mean of the column.

```python
mean_imputer = SimpleImputer(strategy="mean")
df_mean = pd.DataFrame(mean_imputer.fit_transform(df), columns=df.columns)
df_mean
```

## 🔹 Median Imputation

Missing values are replaced with the median of the column.

```python
median_imputer = SimpleImputer(strategy="median")
df_median = pd.DataFrame(median_imputer.fit_transform(df), columns=df.columns)
df_median
```

## 🔹 KNN Imputation

KNN imputation fills missing values using nearest neighbors.

```python
knn_imputer = KNNImputer(n_neighbors=2)
df_knn = pd.DataFrame(knn_imputer.fit_transform(df), columns=df.columns)
df_knn
```

##  Comparison of Methods

| Method  | Pros | Cons |
|--------|------|------|
| Mean   | Simple, fast | Sensitive to outliers |
| Median | Robust to outliers | May ignore distribution |
| KNN    | More accurate | Computationally expensive |

##  Conclusion

- Mean and median imputation are simple and quick methods
- KNN imputation provides better results but is slower
- The choice of method depends on the dataset and problem
