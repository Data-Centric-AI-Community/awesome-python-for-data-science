# Cross Validation in Machine Learning

##  What is Cross Validation?

Cross-validation is a technique used to evaluate how well a machine learning model performs on unseen data.

Instead of splitting the data once into training and testing sets, cross-validation splits the data multiple times and evaluates the model repeatedly.

##  Why do we need Cross Validation?

When we use a simple train-test split:
- The result may depend heavily on how the data is split
- Model may overfit or underfit

Cross-validation solves this by:
- Using multiple splits
- Giving a more reliable performance estimate

##  Dataset Used

We are using the Diabetes dataset from sklearn, which is commonly used for regression problems.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.datasets import load_diabetes

#load dataset
data = load_diabetes()
#features and targets
X = data.data
y = data.target

model = LinearRegression()

scores = cross_val_score(model, X, y, cv=5) #Here, cv=5 means the dataset is split into 5 equal parts. The model is trained and tested 5 times, each time using a different part as test data. 

print("Scores:", scores)
print("Mean Score:", scores.mean())
```

##  Understanding the Results

- Each value in the scores array represents the model's performance on one fold
- The mean score gives an overall performance estimate

A higher score indicates better model performance.
Conclusion: Cross-validation provides a more reliable evaluation of model performance and helps reduce overfitting.

##  Visualization of Cross-Validation Scores

The plot above shows the model performance for each fold in cross-validation.

- Each point represents the score obtained in one fold
- Variation in scores indicates how stable the model is across different data splits
- If the scores are similar, the model is consistent
- Large variation may indicate that the model is sensitive to the data

This visualization helps us better understand the reliability of our model.

```python
import matplotlib.pyplot as plt

plt.plot(scores, marker='o')
plt.title("Cross Validation Scores Across Folds")
plt.xlabel("Fold Number")
plt.ylabel("Score")
plt.grid()
plt.show()
```
